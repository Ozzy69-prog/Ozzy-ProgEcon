```python
Seed = 2026
rng = np.random.default_rng(Seed)  # Random number generator

N = 50_000                      # Number of individuals
Pe = [0.4, 0.35, 0.25]          # probabilities for education
Se = [1, 3, 5]                  # Years of education
He = [1.00, 1.25, 1.55]         # initial human capital
growth = [0.010, 0.020, 0.030]  # growth in human capital
depreciation = 0.06             # Depreciation while employed
std = 0.10                      # Standard deviation of shocks
job_finding = 0.60              # Job finding probability
job_separation = 0.05           # Job separation probability
grant = 0.45                    # Student Grant
replacement = 0.60              # Replacement rate
benefit_floor = 0.35            # Benefit floor
```

```python
short_education = {}
medium_education = {}
long_education = {}

def draw_education(N, Pe):

    for i in range(N):
        education = rng.random()

        if education <= Pe[0]:
            short_education[i] = {
             "education": Se[0],
             "human_capital": He[0],
             "growth": growth[0]
        }
        elif education <= Pe[0] + Pe[1]:
            medium_education[i] = {
                "education": Se[1],
                "human_capital": He[1],
                "growth": growth[1]
        }
        else:
            long_education[i] = {
                "education": Se[2],
                "human_capital": He[2],
                "growth": growth[2]
        }

draw_education(N, Pe)

print(f"Short education: {len(short_education)}")
print(f"Medium education: {len(medium_education)}")
print(f"Long education: {len(long_education)}")
```

```python
E = 1.0
UE = 0.0

start_age = 18
retire_age = 65

def simulate_labor_market(short_education, medium_education, long_education, start_age, retire_age, job_finding, job_separation):

    states = np.full(N, UE)  # Initial state: all unemployed

    for i in short_education:
        start_age = start_age + 1

        while start_age <= retire_age:
            states = UE

            if states == UE:
                if rng.random() <= job_finding:
                    states = 1 # becomes employed
                else:
                    states = 0 # stays unemployed
            else:
                if rng.random() <= job_separation:
                    states = 0 # becomes unemployed
                else:
                    states = 1 # stays employed
        start_age += 1

    for i in medium_education:
        start_age = start_age + 3

        while start_age <= retire_age:
            states = UE

            if states == UE:
                if rng.random() <= job_finding:
                    states = 1 # becomes employed
                else:
                    states = 0 # stays unemployed
            else:
                if rng.random() <= job_separation:
                    states = 0 # becomes unemployed
                else:
                    states = 1 # stays employed
        start_age += 1

    for i in long_education:
        start_age = start_age + 5

        while start_age <= retire_age:
            states = UE

            if states == UE:
                if rng.random() <= job_finding:
                    states = 1 # becomes employed
                else:
                    states = 0 # stays unemployed
            else:
                if rng.random() <= job_separation:
                    states = 0 # becomes unemployed
                else:
                    states = 1 # stays employed
        start_age += 1

    return states

states = simulate_labor_market(short_education, medium_education, long_education, start_age, retire_age, job_finding, job_separation)

print(f"Final states: {states}")
```
